```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/raw/Telco Customer Churn.csv")
df.head()
```

```python
df.info()
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
) 

df = df.dropna(subset=["TotalCharges"])
df.head()
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

df = df.dropna(subset=["TotalCharges"])

print("Missing values:")
print(df.isnull().sum())
```

```python
contract_summary = (
    df.groupby("Contract")
      .agg(
          TotalCustomers=("customerID", "count"),
          ChurnRate=(
              "Churn",
              lambda x: (x == "Yes").mean()
          )
      )
)

display(contract_summary)
```

```python
df["AvgCharge"] = (
    df["TotalCharges"]
    /
    df["tenure"].replace(0, pd.NA)
)

display(
    df[
        ["tenure", "TotalCharges", "AvgCharge"]
    ].head()
)
```

```python
fig, axes = plt.subplots(
    1,
    2,
    figsize=(12, 5)
)

axes[0].hist(
    df["MonthlyCharges"],
    bins=30,
    edgecolor="black"
)

axes[0].set_title(
    "Monthly Charges Distribution"
)

axes[0].set_xlabel(
    "Monthly Charges"
)

axes[0].set_ylabel(
    "Frequency"
)

df["Churn"].value_counts().plot(
    kind="bar",
    ax=axes[1]
)

axes[1].set_title(
    "Customer Churn Counts"
)

axes[1].set_xlabel(
    "Churn"
)

axes[1].set_ylabel(
    "Count"
)

plt.tight_layout()

plt.savefig(
    "../output/figures/m3_q5_history_and_bar_chart.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
pivot_table = pd.pivot_table(
    df,
    index="Contract",
    columns="Churn",
    values="customerID",
    aggfunc="count",
    fill_value=0
)

display(pivot_table)
```

```python
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, df["tenure"].max()],
    labels=[
        "0-12",
        "12-24",
        "24+"
    ]
)

tenure_analysis = (
    df.groupby("TenureGroup")
      .agg(
          Customers=("customerID", "count"),
          ChurnRate=(
              "Churn",
              lambda x: (x == "Yes").mean()
          )
      )
)

display(tenure_analysis)
```

```python
tenure_analysis["ChurnRate"].plot(
    kind="bar",
    figsize=(7, 4)
)

plt.title(
    "Churn Rate by Tenure Group"
)

plt.ylabel(
    "Churn Rate"
)

plt.show()
```

```python
df["ChargeLevel"] = pd.cut(
    df["MonthlyCharges"],
    bins=3,
    labels=[
        "Low",
        "Medium",
        "High"
    ]
)

charge_analysis = (
    df.groupby("ChargeLevel")
      .agg(
          Customers=("customerID", "count"),
          ChurnRate=(
              "Churn",
              lambda x: (x == "Yes").mean()
          )
      )
)

display(charge_analysis)
```

```python
charge_analysis["ChurnRate"].plot(
    kind="bar",
    figsize=(7, 4)
)

plt.title(
    "Churn Rate by Charge Level"
)

plt.ylabel(
    "Churn Rate"
)

plt.show()
```

```python
numerical_df = df[
    [
        "tenure",
        "MonthlyCharges",
        "TotalCharges",
        "AvgCharge"
    ]
]

plt.figure(figsize=(8, 6))

sns.heatmap(
    numerical_df.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title(
    "Correlation Heatmap"
)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="MonthlyCharges"
)

plt.title(
    "Monthly Charges by Churn"
)

plt.show()
```

```python
ml_dataset = pd.get_dummies(
    df,
    drop_first=True
)

print("Shape:")

print(ml_dataset.shape)

display(
    ml_dataset.head()
)
```

```python
ml_dataset.to_csv(
    "../outputs/telco_ml_dataset.csv",
    index=False
)
```
