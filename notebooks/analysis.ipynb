Q3.Apply Min-Max normalization to scale all numerical features to the range [0,1] using vectorized operations.

Để chuẩn hóa một giá trị $x$ cụ thể trong một cột dữ liệu, công thức là:

$$x_{new} = \frac{x - x_{min}}{x_{max} - x_{min}}$$

Trong đó:
* $x_{min}$: Giá trị nhỏ nhất trong cột dữ liệu.
* $x_{max}$: Giá trị lớn nhất trong cột dữ liệu.

**Kết quả:** Sau khi tính toán, giá trị nhỏ nhất cũ sẽ biến thành **0**, và giá trị lớn nhất cũ sẽ biến thành **1**.

```python
# Tìm min là max của từng feature
min_tenure = np.min(tenure_column)
max_tenure = np.max(tenure_column)
min_monthly_charges = np.min(monthly_charges_column)
max_monthly_charges = np.max(monthly_charges_column)
# Do trong cột total charges có giá trị bị thiếu (NaN) sau khi chuyển đổi, nên chúng ta cần bỏ qua các giá trị NaN khi tính min và max
min_total_charges = np.nanmin(total_charges_column)
max_total_charges = np.nanmax(total_charges_column)

print(f"Min Tenure: {min_tenure}, Max Tenure: {max_tenure}")
print(f"Min Monthly Charges: {min_monthly_charges}, Max Monthly Charges: {max_monthly_charges}")
print(f"Min Total Charges: {min_total_charges}, Max Total Charges: {max_total_charges}") 

# Hàm công thức vectorize để chuẩn hóa dữ liệu
def normalize(min, max, column):
    return (column - min) / (max - min) 

normalized_tenure = normalize(min_tenure, max_tenure, tenure_column)
normalized_monthly_charges = normalize(min_monthly_charges, max_monthly_charges, monthly_charges_column)
normalized_total_charges = normalize(min_total_charges, max_total_charges, total_charges_column)

print(f"Normalized Tenure:\n{normalized_tenure}")
print(f"Normalized Monthly Charges:\n{normalized_monthly_charges}")
print(f"Normalized Total Charges:\n{normalized_total_charges}")
```

```python
# Đưa kết quả vào trong mảng
array_normalized = np.array([normalized_tenure, normalized_monthly_charges, normalized_total_charges]).T 

print(f"Normalized Array:\n{array_normalized}")
```

Q6. Implement a fully vectorized function to compute pairwise Euclidean distance between customers using numerical features only (avoid explicit Python loops).

Khoảng cách Euclidean giữa hai vectơ được tính bằng công thức:
$$d(a, b) = \sqrt{\sum (a_i - b_i)^2}$$

Bình phương hai vế để khai triển hằng đẳng thức:
$$d(a, b)^2 = \sum (a_i - b_i)^2 = \sum a_i^2 + \sum b_i^2 - 2 \sum a_i b_i$$

Chuyển đổi sang biểu thức Đại số tuyến tính (vectơ):
* $\sum a_i^2 = \|a\|^2$ (Bình phương độ dài / L2-norm của vectơ $a$)
* $\sum b_i^2 = \|b\|^2$ (Bình phương độ dài / L2-norm của vectơ $b$)
* $\sum a_i b_i = a \cdot b$ (Tích vô hướng - Dot product giữa $a$ và $b$)

$$d(a, b)^2 = \|a\|^2 + \|b\|^2 - 2(a \cdot b)$$

```python
def pairwise_euclidean(matrix):
    sq_norms = np.sum(matrix**2, axis=1, keepdims=True)
    dist_matrix = np.sqrt(sq_norms + sq_norms.T - 2 * np.dot(matrix, matrix.T))
    return dist_matrix
```

```python
print(f"Pairwise Euclidean Distance Matrix:\n{pairwise_euclidean(array_clean)}")
```

Q7. Implement both Min-Max scaling and Z-score normalization from scratch using NumPy and compare their results.

## Công thức toán học
Với mỗi phần tử $x$:
$$z = \frac{x - \mu}{\sigma}$$

Trong đó:
- $\mu$  Giá trị trung bình của cột.
- $\sigma$  Độ lệch chuẩn của cột.

```python
def z_score_normalize(matrix):
    return (matrix - np.mean(matrix, axis=0)) / np.std(matrix, axis=0)
```

```python
print (f"Z-score Normalized Array:\n{z_score_normalize(array_clean)}")
```

## Công thức 
$$\text{Cosine Similarity}(v_1, v_2) = \frac{v_1 \cdot v_2}{\|v_1\| \|v_2\|}$$

```python
def cosine_similarity(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
```

```python
sim = cosine_similarity(array_clean[0], array_clean[1])
print(f"Cosine Similarity between first two customers: {sim}")
```
